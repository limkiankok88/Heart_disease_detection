## Step 2: Basic SQL Queries

### SELECT, FROM, and WHERE

### JOINS (INNER, LEFT, RIGHT, FULL)

#### Create the Second Table

```python
# Create the departments table
cursor.execute('''CREATE TABLE IF NOT EXISTS departments (
    department_name TEXT PRIMARY KEY,
    location TEXT
)''')

# Insert sample department data
departments = [
    ('Sales', 'New York'),
    ('Marketing', 'Los Angeles'),
    ('IT', 'San Francisco'),
    ('HR', 'Chicago'),
    ('Finance', 'Boston')
]

cursor.executemany('''INSERT INTO departments (department_name, location)
                      VALUES (?, ?)''', departments)
conn.commit()
```

#### If you get an error as : `UNIQUE constraint failed: departments.department_name` then execute the code below : 

```python
# Step 1: Delete all rows
cursor.execute('DELETE FROM departments')
conn.commit()

# Step 2: Drop and recreate the table
cursor.execute('DROP TABLE IF EXISTS departments')
conn.commit()
```

#### Inner Join

An inner join returns records that have matching values in both tables.

```python
query_inner_join = '''
SELECT employee.name, employee.age, employee.salary, departments.location
FROM employee
INNER JOIN departments
ON employee.department = departments.department_name
'''

df_inner_join = pd.read_sql_query(query_inner_join, conn)
display("Inner Join:", df_inner_join)
```

#### Left Outer Join

A left outer join returns all records from the left table, and the matched records from the right table. \
If there is no match, NULL values are returned for columns from the right table.

```python
query_left_outer_join = '''
SELECT employee.name, employee.age, employee.salary, departments.location
FROM employee
LEFT OUTER JOIN departments
ON employee.department = departments.department_name
'''

df_left_outer_join = pd.read_sql_query(query_left_outer_join, conn)
display("Left Outer Join:", df_left_outer_join)
```

#### Right Outer Join

**SQLite does not directly support right outer joins.** \
However, simulate a right outer join by swapping the table order in a left join and adjusting the query.

**SQLite version 3.39.0 and later supports RIGHT OUTER JOIN directly**

```python
query_right_outer_join = '''
SELECT employee.name, employee.age, employee.salary, departments.location
FROM departments
LEFT OUTER JOIN employee
ON departments.department_name = employee.department
'''

df_right_outer_join = pd.read_sql_query(query_right_outer_join, conn)
display("Right Outer Join:", df_right_outer_join)
```

#### Full Outer Join

**SQLite does not directly support full outer joins.** \
Simulate a full outer join using a combination of left and right joins and a union operation.

**SQLite version 3.39.0 and later supports FULL OUTER JOIN directly**

```python
query_full_outer_join = '''
SELECT employee.name, employee.age, employee.salary, departments.location
FROM employee
LEFT OUTER JOIN departments
ON employee.department = departments.department_name

UNION

SELECT employee.name, employee.age, employee.salary, departments.location
FROM departments
LEFT OUTER JOIN employee
ON departments.department_name = employee.department
'''

df_full_outer_join = pd.read_sql_query(query_full_outer_join, conn)
display("Full Outer Join:", df_full_outer_join)
```

### GROUP BY, HAVING, and ORDER BY

```python
# Get the total salary by department, ordered by the total salary
query = '''
SELECT department, SUM(salary) AS total_salary
FROM employee
GROUP BY department
HAVING total_salary > 200000
ORDER BY total_salary DESC
'''
pd.read_sql_query(query, conn)
```

### LIMIT

```python
# Select the first 5 employees with the highest salaries
query = '''SELECT * FROM employee ORDER BY salary DESC LIMIT 5'''
pd.read_sql_query(query, conn)
```

## Step 5: Parameterized SQL Queries

```python
# Using parameterized queries to prevent SQL injection
def get_employee_by_department(dept):
    query = '''SELECT * FROM employee WHERE department = ?'''
    return pd.read_sql_query(query, conn, params=(dept,))

# Example usage
get_employee_by_department("IT")
```

## Step 7: Database Security

Preventing SQL injection through parameterized queries has already been demonstrated.

A key security concept: Avoid using raw SQL queries with user input directly.

```python
# Bad practice (vulnerable to SQL injection)
def bad_get_employee_by_name(name):
    query = f"SELECT * FROM employee WHERE name = '{name}'"
    return pd.read_sql_query(query, conn)

# Good practice (parameterized queries)
def secure_get_employee_by_name(name):
    query = "SELECT * FROM employee WHERE name = ?"
    return pd.read_sql_query(query, conn, params=(name,))

# Example of secure query
secure_get_employee_by_name("Sophia Green")
```

The code demonstrates two methods for querying a database to retrieve employee data by name. 

The `bad_get_employee_by_name` function illustrates a vulnerable approach prone to SQL injection attacks, as it directly incorporates the `name` variable into the SQL query string, allowing malicious inputs to manipulate or damage the database (e.g., deleting the table). 

In contrast, the `secure_get_employee_by_name` function employs parameterized queries, a safer practice that prevents SQL injection by using placeholders (`?`) for parameters and passing the actual values separately. This method ensures that user inputs are treated as data rather than executable code, protecting the database from such vulnerabilities.
