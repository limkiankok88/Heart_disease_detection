### Addressing missing, inaccurate, misleading information, data imputation methods

#### Imputation:
For missing data caused by outliers, imputation techniques such as mean, median, or interpolation can be used to replace outlier values with more representative values.

```python
# Impute missing values (if any)
df['id'] = df['id'].bfill()
df['id'] = df['id'].ffill()
#df['id'].fillna(df['id'].mean(), inplace=True) # does not work
df.fillna({'id': df['id'].mean()},inplace=True) #work
df['id'] = df['id'].fillna(df['id'].mean())
display(df.head())

# Replace negative values in 'id' column with NaN
df['id'] = df['id'].apply(lambda x: np.nan if x < 0 else x)

# Interpolation for missing values
df['id'] = df['id'].interpolate(method='linear')

df.isnull().sum()
```

```python
# Check for duplicate records and drop them
data_dup = df[df.duplicated()]
display(data_dup)

df = df.drop_duplicates()

print("\nData after handling duplicate records:")
display(df.head())
```

### Outlier Detection
**Outlier Detection (using zscore Data):**

```python
from scipy import stats

# Calculate z-scores for outlier detection
z_scores = np.abs(stats.zscore(df['weight']))

# Set a threshold for z-score outliers (e.g., z-score > 2)
threshold = 2

# Identify outliers using z-score method
outliers_zscore = df[np.abs(z_scores) > threshold]

print("Data with outliers detected using z-score method:")
print(outliers_zscore)
```

```python
# Replace outliers with mean value
mean_value = df['height'].mean()
df_imputed = df.copy()

df_imputed.loc[abs(z_scores) > threshold, 'height'] = mean_value

print("Data after imputing outliers with mean value:")
print(df_imputed)
```

**IQR Method (Interquartile Range):**\
The IQR method identifies outliers based on the interquartile range, where values outside a specified range are considered outliers.

```python
import pandas as pd

# Calculate quartiles and IQR
Q1 = df['disease'].quantile(0.25)
Q3 = df['disease'].quantile(0.75)
IQR = Q3 - Q1

# Identify outliers using IQR method
outliers_iqr = df[(df['disease'] < (Q1 - 1.5 * IQR)) | (df['disease'] > (Q3 + 1.5 * IQR))]

print("Data with outliers detected using IQR method:")
print(outliers_iqr)
```

**Outlier Management Isolation Forest:**

Let's use Python and `scikit-learn` library to demonstrate outlier management with sample data:

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import IsolationForest

dfx = df.copy()
dfx.drop(columns=['date', 'id', 'country', 'occupation'], inplace=True)

# Apply Isolation Forest for outlier detection
clf = IsolationForest(contamination=0.1, random_state=42)
outlier_labels = clf.fit_predict(dfx)

# Identify and remove outliers
df_cleaned = dfx[outlier_labels != -1]

# Display original and cleaned data
print("Original Data with Outliers:")
display(dfx)

print("\nCleaned Data without Outliers:")
display(df_cleaned)
```

### MinMaxScaler Normalization (Standardization):
**Data Normalization**

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Initialize MinMaxScaler for normalization
scaler = MinMaxScaler()

# Normalize the data using Min-Max scaling
dfx = df.copy()
dfx.drop(columns=['date', 'id', 'country', 'occupation'], inplace=True)

df_normalized = pd.DataFrame(scaler.fit_transform(dfx), columns=dfx.columns) #df.columns

print("Original Data:")
display(dfx)

print("\nNormalized Data:")
display(df_normalized)
```

### Z-Score Normalization (Standardization):
**Z-Score Normalization:**

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Initialize StandardScaler for Z-score normalization
scaler_zscore = StandardScaler()

# Apply Z-score normalization to the data
dfx = df.copy()
dfx.drop(columns=['date', 'id', 'country', 'occupation'], inplace=True)

df_zscore_normalized = pd.DataFrame(scaler_zscore.fit_transform(dfx), columns=dfx.columns)

print("Original Data:")
display(dfx)

print("\nData after Z-score Normalization:")
display(df_zscore_normalized)
```

## Data Validation and Integrity

Validation rules are crucial for enforcing data correctness and consistency. Here's an example of applying clear validation rules using Python code:

```python
import pandas as pd

# Define validation rules
active_rule = df['active'].isin([0, 1])

age_rule = (df['age_group'] >= 2) & (df['age_group'] <= 6)

alco_rule = df['alco'].isin([0, 1])

ap_hi_rule = (df['ap_hi'] >= 59) & (df['ap_hi'] <= 200)
ap_lo_rule = (df['ap_lo'] >= 40) & (df['ap_lo'] <= 200)

cholesterol_rule = df['cholesterol'].isin([1, 2, 3])

gender_rule = df['gender'].isin([1, 2])

gluc_rule = df['gluc'].isin([1, 2, 3])

smoke_rule = df['smoke'].isin([0, 1])
height_rule = (df['height'] >= 0) & (df['height'] <= 300)
weight_rule = (df['weight'] >= 0) & (df['weight'] <= 250)

disease_rule = df['disease'].isin([0, 1])

# Apply validation rules
valid_data = df[active_rule & age_rule & alco_rule & ap_hi_rule & ap_lo_rule & cholesterol_rule & gender_rule & gluc_rule & smoke_rule & height_rule & weight_rule & disease_rule]

# Display valid data after applying validation rules
print("Data after applying validation rules:")
display(valid_data)
```
